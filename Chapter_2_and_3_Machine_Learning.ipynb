## Chapter 2: Supervised Learning
### 1. Sample Datasets & k-NN Classification / Regression

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import mglearn
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier, KNeighborsRegressor
from sklearn.datasets import load_breast_cancer

# Forge dataset
X_forge, y_forge = mglearn.datasets.make_forge()
X_train, X_test, y_train, y_test = train_test_split(X_forge, y_forge, random_state=0)
clf_knn = KNeighborsClassifier(n_neighbors=3).fit(X_train, y_train)
print("k-NN (n_neighbors=3) Forge test predictions:", clf_knn.predict(X_test))
print("k-NN (n_neighbors=3) Forge test accuracy: {:.2f}".format(clf_knn.score(X_test, y_test)))

# Wave dataset regressor
X_wave, y_wave = mglearn.datasets.make_wave(n_samples=40)
X_train_w, X_test_w, y_train_w, y_test_w = train_test_split(X_wave, y_wave, random_state=0)
reg_knn = KNeighborsRegressor(n_neighbors=3).fit(X_train_w, y_train_w)
print("k-NN Regressor (n_neighbors=3) test R^2: {:.2f}".format(reg_knn.score(X_test_w, y_test_w)))
```

### 2. Linear Models (OLS, Ridge, Lasso, Logistic Regression, LinearSVC)

```python
from sklearn.linear_model import LinearRegression, Ridge, Lasso, LogisticRegression
from sklearn.svm import LinearSVC

# Extended Boston dataset
X_boston, y_boston = mglearn.datasets.load_extended_boston()
X_train_b, X_test_b, y_train_b, y_test_b = train_test_split(X_boston, y_boston, random_state=0)

lr_b = LinearRegression().fit(X_train_b, y_train_b)
print("LinearRegression Boston train score: {:.2f}, test score: {:.2f}".format(lr_b.score(X_train_b, y_train_b), lr_b.score(X_test_b, y_test_b)))

ridge = Ridge(alpha=1.0).fit(X_train_b, y_train_b)
print("Ridge (alpha=1.0) Boston train score: {:.2f}, test score: {:.2f}".format(ridge.score(X_train_b, y_train_b), ridge.score(X_test_b, y_test_b)))

lasso001 = Lasso(alpha=0.01, max_iter=100000).fit(X_train_b, y_train_b)
print("Lasso (alpha=0.01) Boston train score: {:.2f}, test score: {:.2f}".format(lasso001.score(X_train_b, y_train_b), lasso001.score(X_test_b, y_test_b)))

cancer = load_breast_cancer()
X_train_c, X_test_c, y_train_c, y_test_c = train_test_split(cancer.data, cancer.target, stratify=cancer.target, random_state=42)
logreg = LogisticRegression(max_iter=5000).fit(X_train_c, y_train_c)
print("LogisticRegression Breast Cancer test score: {:.3f}".format(logreg.score(X_test_c, y_test_c)))
```

### 3. Naive Bayes, Decision Trees, Random Forests, Gradient Boosting

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier

tree_pruned = DecisionTreeClassifier(max_depth=4, random_state=0).fit(X_train_c, y_train_c)
print("Decision Tree (max_depth=4) test acc: {:.3f}".format(tree_pruned.score(X_test_c, y_test_c)))

rf = RandomForestClassifier(n_estimators=100, random_state=0).fit(X_train_c, y_train_c)
print("Random Forest (100 trees) test acc: {:.3f}".format(rf.score(X_test_c, y_test_c)))

gbrt = GradientBoostingClassifier(random_state=0, max_depth=1).fit(X_train_c, y_train_c)
print("Gradient Boosting (max_depth=1) test acc: {:.3f}".format(gbrt.score(X_test_c, y_test_c)))
```

### 4. Support Vector Machines & Multilayer Perceptrons (Neural Networks)

```python
from sklearn.svm import SVC
from sklearn.neural_network import MLPClassifier

# Manual MinMaxScaler for SVM
min_on_training = X_train_c.min(axis=0)
range_on_training = (X_train_c - min_on_training).max(axis=0)
X_train_scaled = (X_train_c - min_on_training) / range_on_training
X_test_scaled = (X_test_c - min_on_training) / range_on_training

svc_scaled = SVC(C=1000).fit(X_train_scaled, y_train_c)
print("Scaled Kernel SVM (C=1000) test acc: {:.3f}".format(svc_scaled.score(X_test_scaled, y_test_c)))

# Standardization for MLP
mean_on_train = X_train_c.mean(axis=0)
std_on_train = X_train_c.std(axis=0)
X_train_mlp = (X_train_c - mean_on_train) / std_on_train
X_test_mlp = (X_test_c - mean_on_train) / std_on_train

mlp = MLPClassifier(max_iter=1000, alpha=1.0, random_state=0).fit(X_train_mlp, y_train_c)
print("Scaled MLP (alpha=1.0) test acc: {:.3f}".format(mlp.score(X_test_mlp, y_test_c)))
```

## Chapter 3: Unsupervised Learning and Preprocessing
### 1. Preprocessing & Scaling

```python
from sklearn.preprocessing import MinMaxScaler, StandardScaler

scaler_mm = MinMaxScaler().fit(X_train_c)
X_train_mm = scaler_mm.transform(X_train_c)
X_test_mm = scaler_mm.transform(X_test_c)

svm_mm = SVC(C=100).fit(X_train_mm, y_train_c)
print("MinMaxScaler + SVC test acc: {:.3f}".format(svm_mm.score(X_test_mm, y_test_c)))

scaler_std = StandardScaler().fit(X_train_c)
X_train_std = scaler_std.transform(X_train_c)
X_test_std = scaler_std.transform(X_test_c)

svm_std = SVC(C=100).fit(X_train_std, y_train_c)
print("StandardScaler + SVC test acc: {:.3f}".format(svm_std.score(X_test_std, y_test_c)))
```

### 2. Dimensionality Reduction (PCA, NMF, t-SNE)

```python
from sklearn.decomposition import PCA, NMF
from sklearn.manifold import TSNE
from sklearn.datasets import load_digits

# PCA on Breast Cancer
pca = PCA(n_components=2).fit(X_train_std)
X_pca = pca.transform(X_train_std)
print("PCA reduced shape:", X_pca.shape)

# NMF on Synthetic Signals
S_signals = mglearn.datasets.make_signals()
A_signals = np.random.RandomState(0).uniform(size=(100, 3))
X_signals = np.dot(S_signals, A_signals.T)
nmf = NMF(n_components=3, random_state=42)
S_nmf = nmf.fit_transform(X_signals)
print("NMF recovered signals shape:", S_nmf.shape)

# t-SNE on Digits
digits = load_digits()
digits_tsne = TSNE(random_state=42).fit_transform(digits.data)
print("Digits t-SNE shape:", digits_tsne.shape)
```

### 3. Clustering (k-Means, Agglomerative, DBSCAN) & Evaluation

```python
from sklearn.cluster import KMeans, AgglomerativeClustering, DBSCAN
from sklearn.metrics.cluster import adjusted_rand_score, silhouette_score
from sklearn.datasets import make_moons

X_moons, y_moons = make_moons(n_samples=200, noise=0.05, random_state=0)
X_moons_scaled = StandardScaler().fit_transform(X_moons)

algorithms = [
    KMeans(n_clusters=2, random_state=0),
    AgglomerativeClustering(n_clusters=2),
    DBSCAN(eps=0.5)
]

for alg in algorithms:
    labels = alg.fit_predict(X_moons_scaled)
    ari = adjusted_rand_score(y_moons, labels)
    sil = silhouette_score(X_moons_scaled, labels)
    print(f"{alg.__class__.__name__:25s} -> ARI: {ari:.2f}, Silhouette Score: {sil:.2f}")
```
